```python
import docker
from inspect import getmembers as gm 
from pprint import pprint
from datetime import datetime
client = docker.from_env()
```

```python
api_container = [x for x in client.containers.list() if x.name.startswith("p2f_api_")][0]
pprint(gm(api_container))
```

```python
created = api_container.attrs["Created"]
created = f'{created.split(".")[0]}Z'
datetime.strptime(created, "%Y-%m-%dT%H:%M:%SZ")
```

```python
pprint(api_container.attrs["Config"]["Env"])
pprint(api_container.attrs["NetworkSettings"])
```

```python
netconfig = api_container.attrs["NetworkSettings"]
networkapi = [x for x in netconfig["Networks"].keys() if x.startswith("p2f_network_test")][0]
print(networkapi)
```
